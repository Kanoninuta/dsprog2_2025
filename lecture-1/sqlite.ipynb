```python
path = ''
db_name = 'car.db'

try:
    conn = sqlite3.connect(path + db_name)

    cur = conn.cursor()

    sql = 'create table cars(id INT, name TEXT, price REAL);'

    cur.execute(sql)

except sqlite3.Error as e:
    print('エラーが発生しました')

finally:
    conn.close()
```

データをDBに保存する（単一レコード）

```python
try:
    path = ''

    db_name = 'car.db'

    conn = sqlite3.connect(path + db_name)

    cur = conn.cursor()

    sql = "INSERT INTO cars(id, name, price) VALUES(1, '370z', 7000000);"

    cur.execute(sql)

    conn.commit()

except sqlite3.Error as e:
    print('エラーが発生しました')

finally:
    conn.close()
```

まとめて入れたい時

```python
try:
    path = ''

    db_name = 'car.db'

    conn = sqlite3.connect(path + db_name)

    cur = conn.cursor()
    # ?→プレースホルダー
    sql = "INSERT INTO cars(id, name, price) VALUES(?, ?, ?);"

    cars = [
        (2, '911カレラ', 15500000),
        (3, 'GRスープラ', 8000000),
        (4, 'シビック Type-R', 5000000)
    ]
    # ここにタプルの名前を入れる

    cur.executemany(sql, cars)

    conn.commit()

except sqlite3.Error as e:
    print('エラーが発生しました')

finally:
    conn.close()
```

```python
try:
    path = ''

    db_name = 'car.db'

    conn = sqlite3.connect(path + db_name)

    cur = conn.cursor()

    sql = "SELECT * FROM cars;"
    cur.execute(sql)

    conn.commit()

except sqlite3.Error as e:
    print('エラーが発生しました')

else:
    for row in cur:
        id, name, price = row
        print(id, name, price)
        
finally:
    conn.close()
```

```python
try:
    path = ''

    db_name = 'car.db'

    conn = sqlite3.connect(path + db_name)

    cur = conn.cursor()

    sql = "DELETE FROM cars WHERE id = 1;"
    # DELETE FROM cars WHERE name like "%カレラ"　曖昧検索
    cur.execute(sql)

    conn.commit()

except sqlite3.Error as e:
    print('エラーが発生しました')


finally:
    conn.close()
```

```python
try:
    path = ''

    db_name = 'car.db'

    conn = sqlite3.connect(path + db_name)

    cur = conn.cursor()

    sql = "UPDATE cars SET price = ? WHERE id = ?;"
    # DELETE FROM cars WHERE name like "%カレラ"　曖昧検索
    cur.execute(sql, (5000000, 2))

    conn.commit()

except sqlite3.Error as e:
    print('エラーが発生しました')


finally:
    conn.close()
```
